```python
# Entering the Pandas Series
# Imports
import numpy as np
import pandas as pd
```

```python
# First we need to convert the Python Lists or Numpy Arrays to Pandas Series
# Pandas series is very efficient for working with Single Columns of the data

# we have random array
rng = np.random.default_rng(1)

numbers_arr = rng.integers(1, 50, 25)
numbers_arr
```

```python
# Converting arr to the Pandas Series
numbers_pd = pd.Series(numbers_arr, name="Random Numbers")

numbers_pd
```

```python
# We have 4 basic properties of the series
print("Values")
numbers_pd.values
```

```python
print("Index")
numbers_pd.index
```

```python
print("Name")
numbers_pd.name
```

```python
print("dtype")
numbers_pd.dtype
```

```python
# We can convert Pandas Series' elements data types

# astype()
print("astype, changing data type to float")
numbers_pd.astype("float")
```

```python
print("astype, changing data type to bool")
numbers_pd.astype("bool")
```
