```python
# Series index

# 1st type of indexing & slicing in Pandas is same as basic Python
sales1 = [0, 5, 155, 0, 588]

sales1 = pd.Series(sales1)

print(sales1)

print()

print("Index: ")
print(sales1[3])

# slice
print()
print("Slice: ")
print(sales1[1:3])

# custom indexes and slices
# create new series
for_index = ["a", "b", "c", "d", "e"]

# we convert sales2 series with custom index

print()
sales2 = [0, 5, 155, 0, 588]
sales2 = pd.Series(sales2, index=for_index, name="Sales 2")

print(sales2)

# we slice the sales2 with our custom indexes
# stopping point is inclusive here, which makes this slicing differs from basic slicing in Python
print()
print("Slicing with custom indexes: ")
print(sales2["b":"d"])
```

```python
# slicing with .iloc method -> .iloc[row position, column position] -> which each of the properties can be like eg, 1 or 2:3

# .iloc method helps us to reach the element by the its propositional index, NOT BY CUSTOM INDEX
print(".iloc method")
print(sales2.iloc[2:4])

# while .loc method is opposite where we can use custom index
print()

print(".loc method")

print(sales2.loc["b":"d"])
```

```python
# while we merged one arr to another for index, it is also possible that we can reset t[he index of the Series to default
# we use .reset_index()

sales2.reset_index()
print(sales2)

# you see that it still not reset, for this we need to make drop=True
print()
print(sales2.reset_index(drop=True))
```

```python
# Sorting in Pandas

# we will sort Series in Pandas using .sort_values() or .sort_index() methods
# .sort_values(ascending=False, inplace=True) -> inplace True sets the original Series to new sorted(like .sort in python)

sales3 = [0, 5, 155, 0, 588]
index_sales3 = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday"]

sales3_series = pd.Series(sales3, index=index_sales3, name="Daily Sales")

sales3_series
```

```python
# sort_values()

print(sales3_series.sort_values())

print()
print(sales3_series.sort_values(ascending=False))
```

```python
# we will sort by index

print(sales3_series.sort_index(ascending=False))
```

```python
# Category Agregation

items_data = ["book 1", "book 2", "book 3", "book 1"] #creating an array

items_sr = pd.Series(items_data, name="Items") #converting it to the Series

print("Series: ", items_sr) #Displaying Series
print()

print("Only unique elements: ", items_sr.unique()) #Displaying only unique elements from the Series
print()

print("Number of unique elements: ", items_sr.nunique()) #Displaying Number of unique items from the Series
print()

print("Number of occurences of elements: ", items_sr.value_counts()) #Displaying number of occurences of elements from the Series
print()

print("Number of occurences of elements: ", items_sr.value_counts(normalize=True)) #Displaying number of occurences of elements from the Series in Percentage
```

```python
# Missing Data Representation in np

missing_item = [100, 103, 200, 30, 1, np.NaN] #np.NaN creates a element space, but missing, no value but treated as float

missing_item_sr = pd.Series(missing_item, name="Missin Item")

print(missing_item_sr) #Series treated as float because of NaN numpy
print()
# example scenario + 2

print(missing_item_sr + 2) #this doesn`t change the value of Missing data
print()
#to handle it correctly for our scenario, we will use numerical operation

print(missing_item_sr.add(2, fill_value=0))
print()

#same scenario in Pd

missing_item2 = pd.Series([100, 103, 200, 30, 1, pd.NA], name="Misser") #np.NaN creates a element space, but missing, no value but treated as float

print(missing_item2) #treated as object data type
```

```python
#identifying missing data

print(missing_item2.isna()) #isna() is method for identifying missing data from column, if missing True, otherwise False
print()

print(missing_item2.isna().value_counts(dropna=False)) #dropna is method of value_counts and making it false help to not to miss the missing data from count
```

```python
#fixing missing data

#these ways are not the standard way of fixing missing data, fixing depends on why is data missing!!!

print(missing_item2.dropna()) #drops the missing data with its corresponding element or index, we can use reset_index() to solve this later
print()

pd.set_option('future.no_silent_downcasting', True) #for future depriciations, not related to the lesson
print(missing_item2.fillna(0))
```

```python
# using custom Python functions in Pandas

missing_item3 = pd.Series([100, 103, 200, 30, 1, pd.NA], name="Misser")

#if we have our own method in Python and want to use it in Pandas, we use .apply() method

def custom_f(checked):
    if pd.isna(checked):
        return "Yes, this is it"
    return "Nope"

print(missing_item3.apply(custom_f))
```

```python
#pd.where method is similar to numpy where method, but it only works with False values

result = missing_item3.where(~missing_item3.isna(), "Yes, this is it").where(missing_item3.isna(), "Nope")
print(result)
```
