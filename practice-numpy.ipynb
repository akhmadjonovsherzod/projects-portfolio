```python
import numpy as np
```

```python
sales = [0, 155, 45, 55, 0, 68, 75, 34, 66, "32"]
sales_array = np.array(sales)

type(sales_array)

# ndim for number of dimensions of numpy array
# usage - object.ndim
print(f"Number of dimensions in Arr: {sales_array.ndim}")

# shape for number of columns for each row(dim)
# usage  object.shape

print(f"Shape of the Arr: {sales_array.shape}")

# size of the array or number of elements in the array
# usage object.size

print(f"Size of the Arr: {sales_array.size}")

# Types of the elements in the array
# usage object.dtype

print(f"Types of the elements in the Arr: {sales_array.dtype}")
```

```python
# ones - generates an array only having 1s in the array
# usage - np.ones((dims, cols), dtype) 

ones_array = np.ones((3, 4), 'str')
print(ones_array)

# zeros - generates an array only having 0s in the array
# usage - np.ones((dims, cols), dtype) 

zeros_array = np.zeros((3, 4), 'int')
print(zeros_array)

# Identity array function makes identity matrix for given size

identity_array = np.identity(10)
print(identity_array)

# reshape method changes the shape of existing array for given size

ones_array.reshape(4, 3)

# linespace method generates an array for given range in a multiple of n

linespc_array = np.linspace(0, 50, 6)
print(linespc_array)
```

```python
# Random number generation

# Numpy random number generator is default_rng function
# import numpy as np
# rng = np.random.default_rng(seed) - seed is like a pool where you can generate same random number with same seed, increases credibility and reusability

rng = np.random.default_rng(616)

my_random_guy = rng.random(10) #random is first mode where generates just numbers between 0 and 1
print(f"rng.random: {my_random_guy}")
print("\n")
my_second = rng.integers(1, 10, 5) #from 1 to 10, 5 integer numbers will be generated
print(f"rng.integers: {my_second}")
print("\n")
my_third = rng.normal(50, 5, 10) #centralised to 50, standard deviation is 5, 10 random numbers
print(f"rng.normal: {my_third}")
```

```python
# assignment reshape

assignment_arr = np.linspace(1, 100, 10)
print(assignment_arr)

reshaped_assignment = assignment_arr.reshape(5, 2)

print(f"Reshaped 5 rows by 2 cols arr: \n {reshaped_assignment} \n")

# assignment random
rng2 = np.random.default_rng(2022)
assignment_random = rng2.integers(1, 100, 9).reshape(3, 3)
print(f"Assignment random: \n {assignment_random}")
```

```python
# slicing the 2 dimensional arrays

# for numpy 1D array slicing and indexing is the same as base Python indexing and slicing, 2D is also, but, one more thing should be added - coordinates
# [row(:), cols(:)]

# usage

slicing_arr = np.arange(1, 37, 1).reshape(6, 6)
print(slicing_arr)

# slicing/indexing

sliced_2d = slicing_arr[:3, :4]
print("\n")
print(sliced_2d)

#  indexed point in 2D
print("\n")
indexed_2d = slicing_arr[1, -1]
print(indexed_2d)
```

```python
# filtering arrays

# for filtering array, we can use logical operations upon array

# 1.Check if the values is 0 or not in the array
print("Check if the values is 0 or not in the array")
array_filter = np.arange(12)
print(array_filter != 0)

# 2. filter by if they are 0, not including them
print()
print("filter by if they are 0, not including them")
print(array_filter[array_filter != 0])

# 3. Multiple logical operations, use | for "or", & for "and". Use a variable to pass the value.

print()
print("Multiple logical operations, use | for 'or', & for 'and'")
filtered_arr = (array_filter != 12) & (array_filter > 5)
print(array_filter[filtered_arr])

# 4. Change all the value larger than 8 to 1
print()
print("Change all the value larger than 8 to 1")
array_filter[array_filter > 8 ] = 1
print(array_filter)
```

```python
# where() for complex filtering

# np.where(condition, result if true, result if false)

arr_where = np.arange(37)
print("My Array for using where")
print(arr_where)

# using where in arr_where by giving condition
print()
print("using where in arr_where by giving condition")
print(np.where(arr_where % 2 == 0, "Divisible", "Non Divisible"))

# complex nested np.where condition
print()
print("complex nested np.where condition")
print(np.where(arr_where % 2 == 0, "Divisible", np.where(arr_where == 5, arr_where, "Other")))
```
